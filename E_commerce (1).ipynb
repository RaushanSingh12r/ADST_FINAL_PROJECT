```python
purchase_summary = df.groupby("purchased").agg(
    total_rows=("purchased", "count"),
    total_revenue=("revenue", "sum"),
    average_revenue=("revenue", "mean")
)

purchase_summary
```

```python
Q1 = df["revenue"].quantile(0.25)
Q3 = df["revenue"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["revenue"] < lower_bound) |
    (df["revenue"] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

print("Total Outliers:", len(outliers))
```

```python
plt.figure(figsize=(10, 5))

sns.boxplot(x=df["revenue"])

plt.title("Revenue Outlier Detection")
plt.xlabel("Revenue")

plt.show()
```

```python
positive_revenue = df[df["revenue"] > 0]["revenue"]

Q1 = positive_revenue.quantile(0.25)
Q3 = positive_revenue.quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

positive_outliers = positive_revenue[
    (positive_revenue < lower_bound) |
    (positive_revenue > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

print("Positive Revenue Records:", len(positive_revenue))

print("Positive Revenue Outliers:",
      len(positive_outliers))
```

```python
plt.figure(figsize=(10, 5))
sns.boxplot(x=positive_revenue)
plt.title("Positive Revenue Outlier Detection")
plt.xlabel("Revenue")
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(
    positive_revenue,
    bins=30,
    kde=True
)

plt.title("Positive Revenue Distribution")
plt.xlabel("Revenue")
plt.ylabel("Frequency")
plt.show()
```

```python
monthly_revenue["period"] = (
    monthly_revenue["year"].astype(str)
    + "-"
    + monthly_revenue["month"].astype(str).str.zfill(2)
)

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_revenue["period"],
    monthly_revenue["revenue"],
    marker="o"
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Year-Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
category_summary = df.groupby("product_category").agg(

    total_orders=("purchased", "sum"),
    total_revenue=("revenue", "sum"),
    average_revenue=("revenue", "mean")

).reset_index()
category_summary = category_summary.sort_values(
    by="total_revenue",
    ascending=False
)
category_summary
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 5))

sns.barplot(
    data=category_summary,
    x="product_category",
    y="total_revenue"
)

plt.title("Total Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.barplot(
    data=category_summary,
    x="product_category",
    y="total_orders"
)
plt.title("Total Orders by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Orders")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
location_summary = df.groupby("location").agg(
    total_orders=("purchased", "sum"),
    total_revenue=("revenue", "sum"),
    average_revenue=("revenue", "mean")

).reset_index()

location_summary = location_summary.sort_values(
    by="total_revenue",
    ascending=False
)
location_summary
```

```python
plt.figure(figsize=(12, 6))
sns.barplot(
    data=location_summary,
    x="location",
    y="total_revenue"
)
plt.title("Total Revenue by Location")
plt.xlabel("Location")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.barplot(
    data=location_summary,
    x="location",
    y="total_orders"
)
plt.title("Total Orders by Location")
plt.xlabel("Location")
plt.ylabel("Total Orders")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
user_type_summary = df.groupby("user_type").agg(
     total_visits=("user_type", "count"),
    total_purchases=("purchased", "sum"),
    total_revenue=("revenue", "sum"),
    average_revenue=("revenue", "mean")
).reset_index()

user_type_summary = user_type_summary.sort_values(
    by="total_purchases",
    ascending=False
)

user_type_summary
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(
    data=user_type_summary,
    x="user_type",
    y="total_purchases"
)

plt.title("Total Purchases by User Type")
plt.xlabel("User Type")
plt.ylabel("Total Purchases")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
device_summary = df.groupby("device_type").agg(
    total_visits=("device_type", "count"),
    total_purchases=("purchased", "sum"),
    total_revenue=("revenue", "sum")
).reset_index()

device_summary = device_summary.sort_values(
    by="total_purchases",
    ascending=False
)

device_summary
```
